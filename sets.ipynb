```python
#creating a set
A={1, 2, 3, 4, 5}
print(A)
B=set()
print(B)
print(type(A))
```

```python
#adding an element to the set
A.add(6)
print(A)
```

```python
#Removing elements from the list
A.remove(4)
print(A)
```

```python
#Union Of Sets
A={10,20,30}
B={5, 15, 25}
print(A.union(B))
```

```python
#Intersection of Two Sets
A={1, 2, 3, 4}
B={3, 4, 5, 6}
print(A.intersection(B))
```

```python
#Difference of two Sets
print(A.difference(B))
```

```python
#Symmetric Difference
print(A.symmetric_difference(B))
```

```python
#Subset
print(A.issubset(B))
```

```python
#set is mutable but set elements are immutable
s={12,'hello',(1,2,3)}
print(s)
```

```python
#update
A.update(s)
print(A)
```

```python
#pop
A.pop()
```

```python
#Superset
print(A.issuperset(s))
```

```python
#Disjoint
print(A.isdisjoint(B))
```

Write a python program that find missing number between 1 and n using a set

```python
n=int(input())
L1=list(map(int,input().split()))
L2=[]
for i in range(1,n+1):
  L2.append(i)
p=set(L1)
q=set(L2)
print(q-p)
```

Write a python program that prints only unique elements in a list

```python
L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)==1:
    L2.append(i)
print(L2)
```
