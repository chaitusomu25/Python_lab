Comparison operators in python like >,<,==,!=,>=,<=
these operators are also known as binary operators because operations takes place between two operators

```python
#Equals to
a=10
b=10
print(a==b)
#Not equals to(!=)
p=10
q=20
print(a!=b)
#Greater than(>)
c=20
d=10
print(c>d)
#Less than(<)
l=5
m=10
print(l<m)
#Greater than or equals to(>=)
f=20
g=20
print(f>=g)
#Less than or equals to(<=)
h=5
i=25
print(h<=i)
```

Logical operators
AND, OR, NOT

```python
#And
a=10
b=20
print(a>5 and b>15)
#OR
print(a>15 or b>15)
#NOT
print(not(a>5))
```

Python identity operators is,is not

```python
#IS
a=[10, 20, 30]
b=a
print(a is b)
#IS NOT
a=[10, 15, 20]
b=[10, 15, 20]
print(a is not b)
```

Membership Operators
in, not in

```python
#IN
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)
#NOT IN
print("orange" not in fruits)
```

Ternary Operators

```python
num = 20
result = "Positive" if num>0 else "Negative"
print(result)
#2nd Example
a=10
b=20
big = a if a>b else b
print(big)
```

Bitwise Operators

```python
a=5
b=3
print(a & b)  #AND
print(a | b)  #OR
print(a^b)    #XOR
print(a<<1)   #Left Shift
print(a>>1)   #Right Shiftm
```

Different ways of printing Output on the console

```python
# 1. Printing a simple message
print("Hello World")

# 2. Printing multiple values
print("Hello", "Python")

# 3. Printing variables
name = "Chaitu"
age = 20
print(name)
print(age)

# 4. Printing using concatenation
print("My name is " + name)

# 5. Printing using comma
print("My name is", name, "and my age is", age)

# 6. Printing using f-string
print(f"My name is {name} and my age is {age}")

# 7. Printing using format()
print("My name is {} and my age is {}".format(name, age))

# 8. Printing with separator
print("Python", "Java", "C", sep="-")

# 9. Printing with end
print("Hello", end=" ")
print("World")

# 10. Printing numbers
print(10, 20, 30)

# 11. Printing expressions
print(10 + 20)

# 12. Printing multiple lines
print("Hello\nWelcome to Python")
```
