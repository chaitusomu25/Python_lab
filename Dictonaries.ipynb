Dictionaries

```python
d = {"name": "Sai", "age": 20, "grade": 9.3}
print(d)
```

```python
#Emty Dictionary
d = {}
print(d)
print(type(d))
```

```python
#Creating dictionary using dict
dic = dict(a=70, m=34)
print(dic)
```

```python
#Accessing Dictionary elements
d = {"name": "Sai", "age": 20}

print(d["name"])
print(d["age"])
```

```python
#printing dictionary keys
d = {"name": "Sai", "age": 20, "grade": 9.3}

for i in d:
    print(i)
```

```python
#printing dictionary values
d = {"name": "Sai", "age": 20, "grade": 9.3}

for i in d:
    print(d[i])
```

```python
#printing keys and values together
d = {"name": "Sai", "age": 20, "grade": 9.3}

for i in d:
    print(i, ":", d[i])
```

```python
#keys method
d = {"name": "Sai", "age": 20, "grade": 9.3}

x = d.keys()
print(x)
print(type(x))
print(list(x))
```

```python
#Values method
d = {"name": "Sai", "age": 20, "grade": 9.3}

y = d.values()
print(y)
print(list(y))
```

```python
#items method
d = {"name": "Sai", "age": 20, "grade": 9.3}

z = d.items()
print(z)
print(list(z))
```

```python
#Adding elements
d = {"name": "Sai", "age": 20}

d["grade"] = 9.3

print(d)
```

```python
#Updating elements
d = {"name": "Sai", "age": 20}

d["age"] = 21

print(d)
```

```python
#Update method
d = {"name": "Sai", "age": 20}

d.update({"grade": 9.3})

print(d)
```

```python
#Pop method
d = {"name": "Sai", "age": 20, "grade": 9.3}

print(d.pop("name"))
print(d)
```

```python
#Pop item method
d = {"name": "Sai", "age": 20, "grade": 9.3}

print(d.popitem())
print(d)
```

```python
#Dictionary comprehension
d = {i: i**2 for i in range(1, 5)}

print(d)
```

```python
#taking input for dictionary keys and values
k = input().split()
v = [int(i) for i in input().split()]

print(k)
print(v)

d = {i: j for i, j in zip(k, v)}

print(d)
```

```python
#iterating using items
d = {"name": "Sai", "age": 20, "grade": 9.3}

for i in d.items():
    print(i)
```

```python
#clear method
d = {"name": "Sai", "age": 20}

d.clear()

print(d)
```

```python
#Copying a dictionary
d = {"name": "Sai", "age": 20}

x = d.copy()

print(x)
```

create a dictionary containing student names and their marks. Print the names and marks of all students and calculate the average marks.

```python
n = int(input("Enter number of students: "))

d = {}

for i in range(n):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    d[name] = marks

print("Student Details:")

for i in d:
    print(i, ":", d[i])

avg = sum(d.values()) / len(d)

print("Average Marks:", avg)
```

count the frequency of each character in a given string using a dictionary.

```python
s = input("Enter a string: ")

d = {}

for i in s:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

print("Character Frequencies:")

for i in d:
    print(i, ":", d[i])
```
