```python
#Methods in Lists
# Creating a list
numbers = [10, 20, 30, 40, 20]

print("Original List:", numbers)

# 1. append() - Adds an element at the end
numbers.append(50)
print("After append:", numbers)

# 2. insert() - Adds an element at a specific position
numbers.insert(1, 15)
print("After insert:", numbers)

# 3. count() - Counts how many times an element occurs
print("Count of 20:", numbers.count(20))

# 4. pop() - Removes an element using its index
numbers.pop(2)
print("After pop:", numbers)

# 5. remove() - Removes a specific element
numbers.remove(20)
print("After remove:", numbers)

# 6. index() - Returns the index of an element
print("Index of 40:", numbers.index(40))

# 7. sort() - Sorts the list in ascending order
numbers.sort()
print("After sort:", numbers)

# 8. extend() - Adds multiple elements to the list
numbers.extend([60, 70, 80])
print("After extend:", numbers)
```

List Comprehension

```python
numbers = [i for i in range(1, 6)]

print(numbers)
```

```python
even = [i for i in range(1, 11) if i % 2 == 0]

print(even)
```

```python
odd = [i for i in range(1, 11) if i % 2 != 0]

print(odd)
```

```python
squares = [i*i for i in range(1, 6)]

print(squares)
```
