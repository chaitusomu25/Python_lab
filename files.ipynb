```python
#creating a new file
file = open("chaitu.txt","w")
content = "This is Chaitanya Lakshmi Somu"
file.write(content)
file.close()
```

```python
#when we wrie the the content into the same file using wite it overwrites the current content
file = open("chaitu.txt","w")
content2 = "currently pursuing my BTech Second year"
file.write(content2)
file.close()
```

```python
#for appending we use append mode
file = open("chaitu.txt","a")
file.write(" at Vishu Institute of technology")
file.close()
```

```python
# for appending in differebt lines we use writelines
file=open("chaitu.txt","a")
file.writelines(["\nChaitanya lakshmi Somu","\n2nd Year","\nVishnu institute of Technology"])
file.close()
```

```python
#for reading we have read mode
with open ("chaitu.txt", "r") as file:
  data = file.read()     #empty arguments read all the data
  print(data)
```

```python
#argument as n reads n characters
with open("chaitu.txt", "r") as file:
  data2 = file.read(100)
  print(data2)
```

```python
# to read one line  we use readline method
with open("chaitu.txt","r") as file:
  data3 = file.readline()
  print(data3)
```

```python
#to read all the sentences as a list elements we have readlines method
with open("chaitu.txt", "r") as file:
  list1 = file.readlines()
  print(list1)
```

```python
#to know the position of cursor we have tell method
file = open("chaitu.txt", "r")
file.readline()
print(file.tell())
```

```python
#to change the position of the cursor we have seek method
file = open("chaitu.txt","r")
file.seek(10)
```
